```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
df = pd.read_csv('../data/dataset_part_2.csv')
df['Year'] = pd.to_datetime(df['Date']).dt.year
```

### Número do voo × Local de lançamento

```python
sns.catplot(y='LaunchSite', x='FlightNumber', hue='Class', data=df, aspect=3)
plt.xlabel('Número do voo'); plt.ylabel('Local de lançamento'); plt.show()
```

### Carga útil × Local de lançamento

```python
sns.catplot(y='LaunchSite', x='PayloadMass', hue='Class', data=df, aspect=3)
plt.xlabel('Carga útil (kg)'); plt.ylabel('Local de lançamento'); plt.show()
```

### Taxa de sucesso por órbita

```python
df.groupby('Orbit')['Class'].mean().sort_values().plot(kind='bar')
plt.ylabel('Taxa de sucesso'); plt.show()
```

### Número do voo × Órbita

```python
sns.catplot(y='Orbit', x='FlightNumber', hue='Class', data=df, aspect=3); plt.show()
```

### Carga útil × Órbita

```python
sns.catplot(y='Orbit', x='PayloadMass', hue='Class', data=df, aspect=3); plt.show()
```

### Tendência anual da taxa de sucesso

```python
df.groupby('Year')['Class'].mean().plot(kind='line', marker='o')
plt.ylabel('Taxa média de sucesso'); plt.show()
```

### Feature engineering (one-hot encoding)

```python
features = df[['FlightNumber','PayloadMass','Orbit','LaunchSite','Flights','GridFins','Reused','Legs','LandingPad','Block','ReusedCount','Serial']]
features_one_hot = pd.get_dummies(features, columns=['Orbit','LaunchSite','LandingPad','Serial']).astype('float64')
features_one_hot.to_csv('../data/dataset_part_3.csv', index=False)
features_one_hot.shape
```
