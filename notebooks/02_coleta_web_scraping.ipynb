```python
import requests, pandas as pd
from bs4 import BeautifulSoup
```

```python
# Revisão fixa da página usada no curso, para resultados reproduzíveis
static_url = "https://en.wikipedia.org/w/index.php?title=List_of_Falcon_9_and_Falcon_Heavy_launches&oldid=1027686922"
response = requests.get(static_url)
soup = BeautifulSoup(response.text, 'html.parser')
soup.title.string
```

```python
html_tables = soup.find_all('table')
first_launch_table = html_tables[2]
column_names = [th.get_text(strip=True) for th in first_launch_table.find_all('th')]
column_names
```

Percorremos as linhas da tabela e guardamos cada campo num dicionário; ajuste a limpeza dos campos conforme o laboratório.

```python
rows = []
for tr in first_launch_table.find_all('tr'):
    cells = [td.get_text(strip=True) for td in tr.find_all('td')]
    if cells:
        rows.append(cells)
df = pd.DataFrame(rows)
df.head()
```

```python
df.to_csv('../data/spacex_web_scraped.csv', index=False)
```
