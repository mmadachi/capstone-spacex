```python
import sqlite3, pandas as pd
con = sqlite3.connect('../data/my_data1.db')
pd.read_csv('../data/Spacex.csv').to_sql('SPACEXTABLE', con, if_exists='replace', index=False)
q = lambda sql: pd.read_sql(sql, con)
```

### 1. Locais de lançamento únicos

```python
q('SELECT DISTINCT Launch_Site FROM SPACEXTABLE')
```

### 2. Cinco registros com local começando por 'CCA'

```python
q("SELECT * FROM SPACEXTABLE WHERE Launch_Site LIKE 'CCA%' LIMIT 5")
```

### 3. Carga total da NASA (CRS)

```python
q("SELECT SUM(PAYLOAD_MASS__KG_) AS total_kg FROM SPACEXTABLE WHERE Customer = 'NASA (CRS)'")
```

### 4. Carga média do F9 v1.1

```python
q("SELECT AVG(PAYLOAD_MASS__KG_) AS media_kg FROM SPACEXTABLE WHERE Booster_Version = 'F9 v1.1'")
```

### 5. Primeiro pouso bem-sucedido em solo

```python
q("SELECT MIN(Date) AS primeira_data FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (ground pad)'")
```

### 6. Drone ship com carga entre 4000 e 6000 kg

```python
q("SELECT DISTINCT Booster_Version FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (drone ship)' AND PAYLOAD_MASS__KG_ > 4000 AND PAYLOAD_MASS__KG_ < 6000")
```

### 7. Missões por resultado

```python
q("SELECT Mission_Outcome, COUNT(*) AS total FROM SPACEXTABLE GROUP BY Mission_Outcome")
```

### 8. Boosters com a carga máxima

```python
q("SELECT DISTINCT Booster_Version FROM SPACEXTABLE WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTABLE)")
```

### 9. Falhas em drone ship em 2015

```python
q("SELECT substr(Date,6,2) AS mes, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTABLE WHERE Landing_Outcome = 'Failure (drone ship)' AND substr(Date,1,4) = '2015'")
```

### 10. Ranking dos resultados de pouso (2010-06-04 a 2017-03-20)

```python
q("SELECT Landing_Outcome, COUNT(*) AS total FROM SPACEXTABLE WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY total DESC")
```
